```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('asd.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
hist = cv2.calcHist([gray], [0], None, [256], [0, 256])
plt.plot(hist)
plt.title('Гистограмма яркости')
plt.xlabel('Яркость')
plt.ylabel('Частота пикселей')
plt.show()
```

```python
dimensions = img.shape
    
print(f"Форма изображения (shape): {dimensions}")
print(f"Высота (высота в пикселях): {dimensions[0]}")
print(f"Ширина (ширина в пикселях): {dimensions[1]}")
```

```python
# Если изображение цветное, у него будет 3 канала (BGR)
if len(dimensions) == 3:
    print(f"Количество каналов цвета: {dimensions[2]}")
```
